```python
# COLAB STEP 14 — Exact V2 architecture
class GCNDrugEncoderV2(nn.Module):
    def __init__(self, in_dim, hidden_dim=384, emb_dim=192, dropout=0.35):
        super().__init__()
        self.conv1 = GCNConv(in_dim, hidden_dim)
        self.bn1 = nn.BatchNorm1d(hidden_dim)
        self.conv2 = GCNConv(hidden_dim, hidden_dim)
        self.bn2 = nn.BatchNorm1d(hidden_dim)
        self.conv3 = GCNConv(hidden_dim, emb_dim)
        self.bn3 = nn.BatchNorm1d(emb_dim)
        self.dropout = dropout

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = self.bn1(x)
        x = F.relu(x)
        x = F.dropout(x, p=self.dropout, training=self.training)

        x = self.conv2(x, edge_index)
        x = self.bn2(x)
        x = F.relu(x)
        x = F.dropout(x, p=self.dropout, training=self.training)

        x = self.conv3(x, edge_index)
        x = self.bn3(x)
        return x

class DDIEdgeClassifierV2(nn.Module):
    def __init__(self, in_dim, num_classes, hidden_dim=512, emb_dim=192, dropout=0.35):
        super().__init__()

        self.encoder = GCNDrugEncoderV2(
            in_dim=in_dim,
            hidden_dim=GCN_HIDDEN_DIM,
            emb_dim=emb_dim,
            dropout=dropout
        )

        pair_dim = emb_dim * 4

        self.classifier = nn.Sequential(
            nn.LayerNorm(pair_dim),
            nn.Linear(pair_dim, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim // 2, num_classes)
        )

    def forward(self, graph_data, d1, d2):
        z = self.encoder(graph_data.x, graph_data.edge_index)
        z1 = z[d1]
        z2 = z[d2]

        pair = torch.cat([
            z1, z2, torch.abs(z1-z2), z1*z2
        ], dim=1)

        return self.classifier(pair)
```

```python
# COLAB STEP 15 — Evaluation
def evaluate_model(model, graph_data, loader, device):
    model.eval()
    all_y, all_pred, all_probs = [], [], []

    with torch.no_grad():
        for d1, d2, y in loader:
            d1 = d1.to(device)
            d2 = d2.to(device)

            logits = model(graph_data, d1, d2)
            probs = torch.softmax(logits, dim=1)

            all_y.append(y.cpu().numpy())
            all_pred.append(probs.argmax(dim=1).cpu().numpy())
            all_probs.append(probs.cpu().numpy())

    y_true = np.concatenate(all_y)
    y_pred = np.concatenate(all_pred)
    probs = np.vstack(all_probs)

    return {
        "accuracy": accuracy_score(y_true, y_pred),
        "macro_f1": f1_score(y_true, y_pred, average="macro", zero_division=0),
        "weighted_f1": f1_score(y_true, y_pred, average="weighted", zero_division=0),
        "y_true": y_true,
        "y_pred": y_pred,
        "probs": probs,
    }
```

```python
# ============================================================
# CONTROLLED ABLATION 2
# PAIR OPERATOR: CONCAT ONLY [z1, z2]
# Everything else = original Weighted V2
# ============================================================

import os
import json
import copy
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.data import Data


# ============================================================
# 1) RESTORE ORIGINAL CLASS WEIGHTING
# ============================================================

class_counts_pair = np.bincount(
    df_train["label_id"].values,
    minlength=num_classes
)

pair_weights_np = np.sqrt(
    class_counts_pair.max() / (class_counts_pair + 1)
)

pair_weights_np = np.clip(
    pair_weights_np,
    0.5,
    3.0
)

pair_weights_np = pair_weights_np / pair_weights_np.mean()

class_weights_tensor = torch.tensor(
    pair_weights_np,
    dtype=torch.float32
).to(device)

print("Class weighting restored ")
print("Min weight:", class_weights_tensor.min().item())
print("Max weight:", class_weights_tensor.max().item())


# ============================================================
# 2) NEW MODEL
# Only change:
# pair = [z1, z2]
# instead of [z1,z2,abs(z1-z2),z1*z2]
# ============================================================

class DDIEdgeClassifierConcatOnly(nn.Module):

    def __init__(
        self,
        in_dim,
        num_classes,
        hidden_dim=512,
        emb_dim=192,
        dropout=0.35
    ):
        super().__init__()

        # SAME GCN encoder as V2
        self.encoder = GCNDrugEncoderV2(
            in_dim=in_dim,
            hidden_dim=GCN_HIDDEN_DIM,
            emb_dim=emb_dim,
            dropout=dropout
        )

        # z1 + z2
        pair_dim = emb_dim * 2   # 192 * 2 = 384

        # SAME classifier design
        self.classifier = nn.Sequential(

            nn.LayerNorm(pair_dim),

            nn.Linear(pair_dim, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.ReLU(),
            nn.Dropout(dropout),

            nn.Linear(
                hidden_dim // 2,
                num_classes
            )
        )

    def forward(self, graph_data, d1, d2):

        z = self.encoder(
            graph_data.x,
            graph_data.edge_index
        )

        z1 = z[d1]
        z2 = z[d2]

        # ====================================================
        # THE ONLY PAIR-OPERATOR CHANGE
        # ====================================================
        pair = torch.cat(
            [z1, z2],
            dim=1
        )

        return self.classifier(pair)


print("Concat-only model ready ")
print("Pair representation = [z1, z2]")
print("Pair dimension = 384")
```

```python
# ============================================================
# TRAINING FUNCTION FOR CONCAT-ONLY ABLATION
# ============================================================

def train_concat_only_auto_resume(
    seed=42,
    exp_name="V2_PAIR_CONCAT_ONLY"
):

    global SAVE_DIR
    global LAST_CKPT_PATH
    global BEST_CKPT_PATH
    global HISTORY_PATH

    set_seed(seed)

    # --------------------------------------------------------
    # Separate folder for this ablation
    # --------------------------------------------------------

    SAVE_DIR = os.path.join(
        PROJECT_DIR,
        "controlled_ablations",
        "pair_operator_concat_only",
        f"seed_{seed}"
    )

    os.makedirs(
        SAVE_DIR,
        exist_ok=True
    )

    LAST_CKPT_PATH = os.path.join(
        SAVE_DIR,
        "LAST_checkpoint.pt"
    )

    BEST_CKPT_PATH = os.path.join(
        SAVE_DIR,
        "BEST_checkpoint.pt"
    )

    HISTORY_PATH = os.path.join(
        SAVE_DIR,
        "training_history.csv"
    )

    # --------------------------------------------------------
    # Save experiment configuration
    # --------------------------------------------------------

    config = {

        "experiment":
            "V2_pair_operator_concat_only",

        "seed":
            seed,

        "pair_operator":
            "[z1,z2]",

        "pair_dim":
            384,

        "baseline_pair_operator":
            "[z1,z2,abs(z1-z2),z1*z2]",

        "baseline_pair_dim":
            768,

        "class_weighting":
            True,

        "feature_dim":
            2249,

        "morgan_bits":
            2048,

        "rdkit_descriptors":
            201,

        "gcn_dims":
            [2249,384,384,192],

        "dropout":
            0.35,

        "batch_size":
            512,

        "optimizer":
            "AdamW",

        "learning_rate":
            7e-4,

        "weight_decay":
            1e-4,

        "max_epochs":
            120,

        "early_stopping_patience":
            18,

        "model_selection":
            "Calibration Macro-F1"
    }

    with open(
        os.path.join(
            SAVE_DIR,
            "ablation_config.json"
        ),
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            config,
            f,
            indent=2
        )

    # --------------------------------------------------------
    # Data loaders
    # --------------------------------------------------------

    train_loader, cal_loader, test_loader = make_loaders(
        seed
    )

    graph_data = Data(

        x=torch.tensor(
            x_morgan_rdkit,
            dtype=torch.float32
        ),

        edge_index=edge_index

    ).to(device)

    # --------------------------------------------------------
    # CONCAT-ONLY MODEL
    # --------------------------------------------------------

    model = DDIEdgeClassifierConcatOnly(

        in_dim=x_morgan_rdkit.shape[1],

        num_classes=num_classes,

        hidden_dim=CLASSIFIER_HIDDEN_DIM,

        emb_dim=EMB_DIM,

        dropout=DROPOUT

    ).to(device)

    # SAME optimizer
    optimizer = torch.optim.AdamW(

        model.parameters(),

        lr=LEARNING_RATE,

        weight_decay=WEIGHT_DECAY
    )

    # SAME scheduler
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(

        optimizer,

        mode="max",

        factor=SCHEDULER_FACTOR,

        patience=SCHEDULER_PATIENCE
    )

    # SAME weighted CrossEntropy
    criterion = nn.CrossEntropyLoss(
        weight=class_weights_tensor
    )

    start_epoch = 1

    best_cal_macro = -np.inf
    best_epoch = None

    wait = 0

    history = []

    # ========================================================
    # AUTO RESUME
    # ========================================================

    if os.path.exists(LAST_CKPT_PATH):

        ckpt = torch.load(
            LAST_CKPT_PATH,
            map_location=device,
            weights_only=False
        )

        if ckpt.get(
            "input_checksums"
        ) != input_checksums:

            raise RuntimeError(
                "Checkpoint/data mismatch."
            )

        model.load_state_dict(
            ckpt["model_state"]
        )

        optimizer.load_state_dict(
            ckpt["optimizer_state"]
        )

        scheduler.load_state_dict(
            ckpt["scheduler_state"]
        )

        start_epoch = (
            int(ckpt["epoch"]) + 1
        )

        best_cal_macro = float(
            ckpt["best_cal_macro"]
        )

        best_epoch = ckpt["best_epoch"]

        wait = int(
            ckpt["wait"]
        )

        history = list(
            ckpt.get(
                "history",
                []
            )
        )

        print(
            f"Resuming from epoch {start_epoch}"
        )

        if ckpt.get(
            "completed",
            False
        ):

            print(
                "This run is already completed."
            )

            return

    else:

        print(
            "Starting CONCAT-ONLY run from epoch 1."
        )

    # ========================================================
    # TRAIN
    # ========================================================

    try:

        for epoch in range(
            start_epoch,
            MAX_EPOCHS + 1
        ):

            model.train()

            total_loss = 0.0

            for d1, d2, y in tqdm(
                train_loader,
                desc=f"Epoch {epoch}/{MAX_EPOCHS}",
                leave=True
            ):

                d1 = d1.to(device)
                d2 = d2.to(device)
                y = y.to(device)

                optimizer.zero_grad(
                    set_to_none=True
                )

                logits = model(
                    graph_data,
                    d1,
                    d2
                )

                loss = criterion(
                    logits,
                    y
                )

                loss.backward()

                torch.nn.utils.clip_grad_norm_(
                    model.parameters(),
                    GRAD_CLIP_NORM
                )

                optimizer.step()

                total_loss += (
                    loss.item()
                )

            avg_loss = (
                total_loss /
                len(train_loader)
            )

            cal_res = evaluate_model(
                model,
                graph_data,
                cal_loader,
                device
            )

            scheduler.step(
                cal_res["macro_f1"]
            )

            current_lr = (
                optimizer
                .param_groups[0]["lr"]
            )

            improved = (
                cal_res["macro_f1"]
                >
                best_cal_macro
            )

            if improved:

                best_cal_macro = float(
                    cal_res["macro_f1"]
                )

                best_epoch = epoch

                wait = 0

            else:

                wait += 1

            row = {

                "experiment":
                    exp_name,

                "seed":
                    seed,

                "epoch":
                    epoch,

                "loss":
                    avg_loss,

                "cal_accuracy":
                    cal_res["accuracy"],

                "cal_macro_f1":
                    cal_res["macro_f1"],

                "cal_weighted_f1":
                    cal_res["weighted_f1"],

                "lr":
                    current_lr,

                "wait":
                    wait,

                "improved":
                    improved
            }

            history.append(row)

            save_history_atomic(
                history,
                HISTORY_PATH
            )

            # BEST
            if improved:

                save_atomic_torch(

                    make_payload(

                        model,
                        optimizer,
                        scheduler,

                        epoch,

                        best_cal_macro,

                        best_epoch,

                        wait,

                        history,

                        seed
                    ),

                    BEST_CKPT_PATH
                )

                print(
                    " BEST checkpoint saved."
                )

            # LAST
            save_atomic_torch(

                make_payload(

                    model,
                    optimizer,
                    scheduler,

                    epoch,

                    best_cal_macro,

                    best_epoch,

                    wait,

                    history,

                    seed
                ),

                LAST_CKPT_PATH
            )

            print(

                f"Epoch {epoch:03d} | "

                f"Loss={avg_loss:.4f} | "

                f"Cal Acc="
                f"{cal_res['accuracy']:.4f} | "

                f"Cal Macro="
                f"{cal_res['macro_f1']:.4f} | "

                f"Cal Weighted="
                f"{cal_res['weighted_f1']:.4f} | "

                f"LR={current_lr:.1e} | "

                f"Wait="
                f"{wait}/{EARLY_STOP_PATIENCE} | "

                f"BestEpoch={best_epoch} | "

                f"BestMacro="
                f"{best_cal_macro:.4f}"
            )

            if wait >= EARLY_STOP_PATIENCE:

                print(
                    f"Early stopping at epoch {epoch}."
                )

                break

    except KeyboardInterrupt:

        print(
            "\nTraining interrupted."
        )

        print(
            "Last completed epoch is saved."
        )

        return

    # ========================================================
    # LOAD TRUE BEST
    # ========================================================

    best_ckpt = torch.load(

        BEST_CKPT_PATH,

        map_location=device,

        weights_only=False
    )

    model.load_state_dict(
        best_ckpt["model_state"]
    )

    cal_final = evaluate_model(

        model,
        graph_data,
        cal_loader,
        device
    )

    test_final = evaluate_model(

        model,
        graph_data,
        test_loader,
        device
    )

    gap = abs(

        cal_final["macro_f1"]

        -

        best_ckpt["best_cal_macro"]
    )

    print(
        "\nFINAL RESULT FROM TRUE BEST CHECKPOINT"
    )

    print(
        "Best epoch:",
        best_ckpt["best_epoch"]
    )

    print(
        "Saved best CAL Macro:",
        best_ckpt["best_cal_macro"]
    )

    print(
        "Reloaded CAL Macro:",
        cal_final["macro_f1"]
    )

    print(
        "Checkpoint gap:",
        gap
    )

    print(
        "TEST Acc:",
        test_final["accuracy"]
    )

    print(
        "TEST Macro:",
        test_final["macro_f1"]
    )

    print(
        "TEST Weighted:",
        test_final["weighted_f1"]
    )

    # --------------------------------------------------------
    # Save final result
    # --------------------------------------------------------

    result = {

        "experiment":
            exp_name,

        "seed":
            seed,

        "pair_operator":
            "concat_only",

        "best_epoch":
            best_ckpt["best_epoch"],

        "cal_accuracy":
            cal_final["accuracy"],

        "cal_macro_f1":
            cal_final["macro_f1"],

        "cal_weighted_f1":
            cal_final["weighted_f1"],

        "test_accuracy":
            test_final["accuracy"],

        "test_macro_f1":
            test_final["macro_f1"],

        "test_weighted_f1":
            test_final["weighted_f1"],

        "checkpoint_gap":
            gap
    }

    pd.DataFrame(
        [result]
    ).to_csv(

        os.path.join(
            SAVE_DIR,
            "final_result.csv"
        ),

        index=False
    )

    # mark completed
    save_atomic_torch(

        make_payload(

            model,
            optimizer,
            scheduler,

            history[-1]["epoch"],

            best_cal_macro,

            best_epoch,

            wait,

            history,

            seed,

            completed=True
        ),

        LAST_CKPT_PATH
    )

    return result
```
